```python
#linear regrission: used for predicting continous values
#single linear regrission:one independent value
#multiple linear regression:one or more independent values
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# Sample House Price Dataset
data = pd.DataFrame({
    "Size(sqft)": [1200, 1500, 800, 2000, 1100, 1800, 950, 1600, 2200, 1400],
    "Bedrooms": [3, 4, 2, 5, 3, 4, 2, 3, 5, 3],
    "Age": [5, 2, 10, 1, 8, 3, 12, 4, 1, 6],
    "LocationScore": [8, 9, 6, 10, 7, 9, 5, 8, 10, 7],
    "Price": [50000, 70000, 30000, 100000, 45000, 75000, 28000, 68000, 120000, 60000]
})
print(data)
```

```python
x=data[['Size(sqft)','Bedrooms','Age','LocationScore']]
print(x)
```

```python
y=data['Price']
print(y)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
X_train
```

```python
Y_train
```

```python
model.fit(X_train,Y_train)
```

```python
predict=model.predict(X_test)
predict
```

```python
from sklearn.metrics import mean_absolute_error, r2_score
print("MAE:", mean_absolute_error(Y_test, predict))
print("R2 Score:", r2_score(Y_test, predict))
```

```python
new_house = [[1200,3,5,8]]
predicted_price = model.predict(new_house)
print("Predicted Price:", predicted_price[0])
```

```python
plt.scatter(data['Size(sqft)'], data['Price'], color='blue')
plt.xlabel("Area (sqft)")
plt.ylabel("Price")
plt.title("House Price Prediction")
plt.show()
```
